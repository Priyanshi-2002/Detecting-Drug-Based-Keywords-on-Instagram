```python
# Import necessary libraries
import pandas as pd
import re
from nltk.corpus import words, stopwords

# Load the dataset (Replace with the actual file path)
df = pd.read_csv(r"C:\Users\Priyanshi Tomar\Desktop\dataset_instagram-hashtag-scraper_2024-12-20_18-44-10-941.csv", encoding="ISO-8859-1")

# Print column names to verify dataset
print(df.columns)

# Load English dictionary and stopwords
english_words = set(words.words())
stop_words = set(stopwords.words('english'))

# Define text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)  # Remove URLs
    text = re.sub(r'\W+', ' ', text)  # Remove special characters
    text = ' '.join(word for word in text.split() if word not in stop_words and word not in english_words)
    return text

# Combine caption and hashtag columns and clean the text
df['text'] = (df['caption'].fillna('') + ' ' + df['hashtag'].fillna('')).apply(clean_text)

# Drop original columns
df = df.drop(columns=['caption', 'hashtag'])

# Save the cleaned dataset to a new file
output_path = r"C:\Users\Priyanshi Tomar\Desktop\trial.csv"
df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved at: {output_path}")
```
