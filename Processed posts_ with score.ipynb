```python
import pandas as pd

# Load the keyword score CSV (Fuzzfinal.csv)
keyword_scores = pd.read_csv(r"C:\Users\Priyanshi Tomar\Desktop\Fuzzfinal.csv")
high_score_keywords = keyword_scores[keyword_scores['Score'] > 0.75]['Keyword'].str.lower().tolist()

# Load the second CSV with grouped keywords in posts (Trials.csv)
posts = pd.read_csv(r"C:\Users\Priyanshi Tomar\Desktop\trial.csv", encoding='unicode_escape')

# Function to process a single post
def process_post(post_text, high_score_keywords):
    if not isinstance(post_text, str):  # Handle NaN or non-string cases
        return [], 0
    
    # Split post into unique words (ignoring repetitions)
    unique_words = list(set(word.strip().lower() for word in post_text.split()))
    
    # Match unique words with high-score keywords
    matched_keywords = [word for word in unique_words if word in high_score_keywords]
    
    return matched_keywords, len(matched_keywords)

# Apply the function to process each post
posts['High Scoring Keywords'], posts['High Scoring Count'] = zip(
    *posts['text'].apply(lambda x: process_post(x, high_score_keywords))
)

# Save the updated data to a CSV
output_path = r"C:\Users\Priyanshi Tomar\Desktop\processed_posts.csv"
posts.to_csv(output_path, index=False)

print(f"Processed posts data saved to '{output_path}'")
```
